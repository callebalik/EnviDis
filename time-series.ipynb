```python
# Test for overdispersion in count data
import numpy as np
import scipy.stats as stats
from scipy.stats import poisson, nbinom
import matplotlib.pyplot as plt

print("Overdispersion Testing for Count Data")
print("=" * 50)

# Test overdispersion for each count variable
count_variables = ['document_count', 'dis_count', 'pnm_count', 'co_count']
count_names = ['Abstract Count', 'DIS Count', 'PNM Count', 'DPCO Count']

def test_overdispersion(data, name):
    """
    Test for overdispersion using multiple methods
    """
    print(f"\n{name}:")
    print("-" * 30)

    # Basic statistics
    mean_val = np.mean(data)
    var_val = np.var(data, ddof=1)  # Sample variance

    print(f"Mean: {mean_val:.2f}")
    print(f"Variance: {var_val:.2f}")
    print(f"Variance-to-mean ratio: {var_val/mean_val:.2f}")

    # Interpretation of variance-to-mean ratio
    if var_val/mean_val > 1.2:
        dispersion_status = "Overdispersed"
    elif var_val/mean_val < 0.8:
        dispersion_status = "Underdispersed"
    else:
        dispersion_status = "Equidispersed (Poisson-like)"

    print(f"Dispersion status: {dispersion_status}")

    # Test against Poisson distribution
    # Using Kolmogorov-Smirnov test
    # First, we need to estimate lambda (= mean for Poisson)
    lambda_est = mean_val

    # Generate expected Poisson probabilities
    max_val = int(np.max(data))
    expected_poisson = [poisson.pmf(k, lambda_est) for k in range(max_val + 1)]

    # Calculate observed frequencies
    observed_counts = np.bincount(data.astype(int), minlength=max_val + 1)
    observed_freq = observed_counts / len(data)

    # Chi-square goodness of fit test (for large samples)
    if len(data) > 30 and mean_val > 5:
        expected_counts = np.array(expected_poisson) * len(data)
        # Combine small expected frequencies
        mask = expected_counts >= 5
        if np.sum(mask) >= 3:  # Need at least 3 categories
            obs_combined = np.sum(observed_counts[mask])
            exp_combined = np.sum(expected_counts[mask])

            chi2_stat = np.sum((observed_counts[mask] - expected_counts[mask])**2 / expected_counts[mask])
            df = np.sum(mask) - 1 - 1  # -1 for estimated parameter
            if df > 0:
                p_value_chi2 = 1 - stats.chi2.cdf(chi2_stat, df)
                print(f"Chi-square test vs Poisson: χ² = {chi2_stat:.4f}, p = {format_p_value(p_value_chi2)}")

    # Alternative: Compare with theoretical Poisson using Anderson-Darling or other tests
    # For large counts, we can use normal approximation
    if mean_val > 10:
        # Normal approximation to Poisson
        standardized = (data - mean_val) / np.sqrt(mean_val)
        _, p_normal = stats.normaltest(standardized)
        print(f"Normal test on standardized data: p = {format_p_value(p_normal)}")

    # Index of dispersion test (for time series data)
    # This tests if variance significantly differs from mean
    n = len(data)
    test_stat = (n - 1) * var_val / mean_val
    # Under null hypothesis (Poisson), this follows chi-square with n-1 df
    p_dispersion = 2 * min(stats.chi2.cdf(test_stat, n-1),
                          1 - stats.chi2.cdf(test_stat, n-1))
    print(f"Index of dispersion test: D = {test_stat:.4f}, p = {format_p_value(p_dispersion)}")

    if p_dispersion < 0.05:
        print("→ Significant evidence against Poisson distribution")
    else:
        print("→ No significant evidence against Poisson distribution")

    return var_val/mean_val, p_dispersion

# Test each variable
dispersion_results = {}
for var, name in zip(count_variables, count_names):
    data = yearly_counts[var].values
    ratio, p_val = test_overdispersion(data, name)
    dispersion_results[var] = {'ratio': ratio, 'p_value': p_val, 'name': name}

# Summary table
print("\n" + "=" * 60)
print("Overdispersion Summary")
print("=" * 60)
print(f"{'Variable':<20} {'Var/Mean Ratio':<15} {'p-value':<15} {'Status':<20}")
print("-" * 60)

for var, result in dispersion_results.items():
    ratio = result['ratio']
    p_val = result['p_value']
    name = result['name']

    if ratio > 1.5:
        status = "Highly overdispersed"
    elif ratio > 1.2:
        status = "Moderately overdispersed"
    elif ratio < 0.8:
        status = "Underdispersed"
    else:
        status = "Equidispersed"

    if p_val < 0.05:
        status += " *"

    print(f"{name:<20} {ratio:>14.2f} {format_p_value(p_val):>14} {status:<20}")

print("\n* = Statistically significant deviation from Poisson")

# Visual inspection with histograms
print("\n" + "=" * 60)
print("Visual Inspection of Distributions")
print("=" * 60)

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.ravel()

for i, (var, name) in enumerate(zip(count_variables, count_names)):
    data = yearly_counts[var].values
    mean_val = np.mean(data)

    # Plot histogram
    axes[i].hist(data, bins=min(20, len(np.unique(data))), alpha=0.7, density=True,
                label='Observed', edgecolor='black', linewidth=0.5)

    # Overlay theoretical Poisson (if reasonable)
    if mean_val <= 50:  # Only for reasonable Poisson range
        x_range = np.arange(0, int(np.max(data)) + 1)
        poisson_probs = poisson.pmf(x_range, mean_val)
        axes[i].plot(x_range, poisson_probs, 'ro-', alpha=0.7,
                    label=f'Poisson(λ={mean_val:.1f})', markersize=3)

    axes[i].set_title(f'{name}\nVar/Mean = {dispersion_results[var]["ratio"]:.2f}')
    axes[i].set_xlabel('Count')
    axes[i].set_ylabel('Density/Probability')
    axes[i].legend()
    axes[i].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Test for negative binomial fit (alternative to Poisson for overdispersed data)
print("\n" + "=" * 60)
print("Negative Binomial Model Comparison")
print("=" * 60)

def fit_negative_binomial(data):
    """Fit negative binomial distribution using method of moments"""
    mean_val = np.mean(data)
    var_val = np.var(data, ddof=1)

    if var_val <= mean_val:
        return None, None  # Cannot fit NB when underdispersed

    # Method of moments estimators
    p = mean_val / var_val
    r = mean_val * p / (1 - p)

    return r, p

for var, name in zip(count_variables, count_names):
    data = yearly_counts[var].values
    ratio = dispersion_results[var]['ratio']

    print(f"\n{name}:")

    if ratio > 1.2:  # Only for overdispersed data
        r, p = fit_negative_binomial(data)
        if r is not None and r > 0:
            print(f"  Negative Binomial parameters: r={r:.4f}, p={p:.4f}")
            print(f"  NB mean: {r*(1-p)/p:.2f}, NB variance: {r*(1-p)/p**2:.2f}")
            print(f"  → Negative Binomial model may be appropriate")
        else:
            print(f"  → Could not fit Negative Binomial distribution")
    else:
        print(f"  → Poisson distribution appears adequate (ratio = {ratio:.2f})")

# Conclusion
print("\n" + "=" * 60)
print("Conclusions")
print("=" * 60)

overdispersed_vars = [var for var, result in dispersion_results.items()
                     if result['ratio'] > 1.2 and result['p_value'] < 0.05]

if overdispersed_vars:
    print("Variables showing significant overdispersion:")
    for var in overdispersed_vars:
        name = dispersion_results[var]['name']
        ratio = dispersion_results[var]['ratio']
        print(f"  • {name} (Var/Mean = {ratio:.2f})")

    print("\nRecommendations:")
    print("  • Consider negative binomial regression instead of Poisson")
    print("  • Use robust standard errors in trend analysis")
    print("  • Consider zero-inflated models if many zeros present")
else:
    print("No strong evidence of overdispersion in the count data.")
    print("Poisson-based models may be appropriate.")
```
