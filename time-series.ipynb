```python
# Negative Binomial Regression for DPCO Counts - Setup and Data Preparation
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Define format_p_value function for consistent formatting
def format_p_value(p):
    """Format p-values consistently"""
    if p < 0.001:
        return "< 0.001"
    elif p < 0.01:
        return f"{p:.3f}"
    else:
        return f"{p:.3f}"

# Try importing statsmodels for negative binomial regression
try:
    import statsmodels.api as sm
    from scipy import stats
    statsmodels_available = True
except ImportError:
    print("Warning: statsmodels not available. Installing or using alternative approach.")
    statsmodels_available = False

print("Negative Binomial Regression Analysis for DPCO Counts")
print("=" * 60)

# Prepare data for modeling
modeling_data = yearly_counts.copy()

# Create time variables
modeling_data['years_since_start'] = modeling_data['year'] - modeling_data['year'].min()
modeling_data['year_centered'] = modeling_data['year'] - modeling_data['year'].mean()

# Create log-transformed mention predictors (add small constant to avoid log(0))
modeling_data['log_dis_count'] = np.log(modeling_data['dis_count'] + 1)
modeling_data['log_pnm_count'] = np.log(modeling_data['pnm_count'] + 1)
modeling_data['log_document_count'] = np.log(modeling_data['document_count'] + 1)

# Check for any issues with the data
print("Data preparation:")
print(f"Sample size: {len(modeling_data)}")
print(f"DPCO count range: {modeling_data['co_count'].min():.0f} to {modeling_data['co_count'].max():.0f}")
print(f"Mean DPCO count: {modeling_data['co_count'].mean():.2f}")
print(f"Variance DPCO count: {modeling_data['co_count'].var():.2f}")
print(f"Variance/Mean ratio: {modeling_data['co_count'].var()/modeling_data['co_count'].mean():.2f}")

# Prepare response variable
y = modeling_data['co_count'].astype(int)
```

```python
# Model 1: Time-only model
if statsmodels_available:
    print("Model 1: Time-only model")
    print("=" * 40)

    X1 = sm.add_constant(modeling_data['years_since_start'])

    try:
        nb_model1 = sm.GLM(y, X1, family=sm.families.NegativeBinomial()).fit()
        print(nb_model1.summary().tables[1])

        # Calculate trend information
        time_coef1 = nb_model1.params['years_since_start']
        time_pval1 = nb_model1.pvalues['years_since_start']
        annual_change_rate1 = (np.exp(time_coef1) - 1) * 100

        print(f"\nTime trend interpretation:")
        print(f"  Coefficient: {time_coef1:.6f}")
        print(f"  Annual change rate: {annual_change_rate1:.2f}% per year")
        print(f"  p-value: {format_p_value(time_pval1)}")

        model1_aic = nb_model1.aic

    except Exception as e:
        print(f"Model 1 failed: {e}")
        nb_model1 = None
else:
    print("Statsmodels not available - skipping Model 1")
    nb_model1 = None
```

```python
# Model 2: Time + log(DIS) + log(PNM)
if statsmodels_available:
    print("Model 2: Time + log(DIS) + log(PNM)")
    print("=" * 40)

    X2 = modeling_data[['years_since_start', 'log_dis_count', 'log_pnm_count']]
    X2 = sm.add_constant(X2)

    try:
        nb_model2 = sm.GLM(y, X2, family=sm.families.NegativeBinomial()).fit()
        print(nb_model2.summary().tables[1])

        # Extract key results
        time_coef2 = nb_model2.params['years_since_start']
        time_pval2 = nb_model2.pvalues['years_since_start']
        dis_coef = nb_model2.params['log_dis_count']
        dis_pval = nb_model2.pvalues['log_dis_count']
        pnm_coef = nb_model2.params['log_pnm_count']
        pnm_pval = nb_model2.pvalues['log_pnm_count']

        annual_change_rate2 = (np.exp(time_coef2) - 1) * 100

        print(f"\nModel 2 interpretation:")
        print(f"  Time coefficient: {time_coef2:.6f} (p = {format_p_value(time_pval2)})")
        print(f"  Annual change rate: {annual_change_rate2:.2f}% per year")
        print(f"  DIS elasticity: {dis_coef:.3f} (p = {format_p_value(dis_pval)})")
        print(f"  PNM elasticity: {pnm_coef:.3f} (p = {format_p_value(pnm_pval)})")

        model2_aic = nb_model2.aic

    except Exception as e:
        print(f"Model 2 failed: {e}")
        nb_model2 = None
else:
    print("Statsmodels not available - skipping Model 2")
    nb_model2 = None
```

```python
# Model 3: Comprehensive model with document count
if statsmodels_available:
    print("Model 3: Time + log(DIS) + log(PNM) + log(Documents)")
    print("=" * 40)

    X3 = modeling_data[['years_since_start', 'log_dis_count', 'log_pnm_count', 'log_document_count']]
    X3 = sm.add_constant(X3)

    try:
        nb_model3 = sm.GLM(y, X3, family=sm.families.NegativeBinomial()).fit()
        print(nb_model3.summary().tables[1])

        time_coef3 = nb_model3.params['years_since_start']
        time_pval3 = nb_model3.pvalues['years_since_start']
        annual_change_rate3 = (np.exp(time_coef3) - 1) * 100

        print(f"\nModel 3 interpretation:")
        print(f"  Time coefficient: {time_coef3:.6f} (p = {format_p_value(time_pval3)})")
        print(f"  Annual change rate: {annual_change_rate3:.2f}% per year")

        model3_aic = nb_model3.aic

    except Exception as e:
        print(f"Model 3 failed: {e}")
        nb_model3 = None
else:
    print("Statsmodels not available - skipping Model 3")
    nb_model3 = None
```

```python
# Model 4: Time with Document Count as Offset (Rate Model)
if statsmodels_available:
    print("Model 4: Time with Document Count as Offset (Rate Model)")
    print("=" * 40)

    X4 = sm.add_constant(modeling_data['years_since_start'])
    offset = modeling_data['log_document_count']

    try:
        nb_model4 = sm.GLM(y, X4, family=sm.families.NegativeBinomial(), offset=offset).fit()
        print(nb_model4.summary().tables[1])

        time_coef4 = nb_model4.params['years_since_start']
        time_pval4 = nb_model4.pvalues['years_since_start']
        annual_change_rate4 = (np.exp(time_coef4) - 1) * 100

        print(f"\nModel 4 interpretation:")
        print(f"  Time coefficient: {time_coef4:.6f} (p = {format_p_value(time_pval4)})")
        print(f"  Annual change rate: {annual_change_rate4:.2f}% per year")
        print(f"  This models the DPCO rate per abstract directly")
        print(f"  Offset coefficient fixed at 1.0 (log(documents) coefficient)")

        model4_aic = nb_model4.aic

    except Exception as e:
        print(f"Model 4 failed: {e}")
        nb_model4 = None
else:
    print("Statsmodels not available - skipping Model 4")
    nb_model4 = None
```

```python
# Model Comparison
if statsmodels_available:
    print("Model Comparison")
    print("=" * 60)

    models = []
    if nb_model1 is not None:
        models.append(("Model 1 (Time only)", nb_model1, model1_aic))
    if nb_model2 is not None:
        models.append(("Model 2 (Time + DIS + PNM)", nb_model2, model2_aic))
    if nb_model3 is not None:
        models.append(("Model 3 (Full model)", nb_model3, model3_aic))
    if nb_model4 is not None:
        models.append(("Model 4 (Rate model)", nb_model4, model4_aic))

    if models:
        print(f"{'Model':<25} {'AIC':<10} {'Log-Likelihood':<15} {'Deviance':<10}")
        print("-" * 60)
        for name, model, aic in models:
            print(f"{name:<25} {aic:<10.2f} {model.llf:<15.2f} {model.deviance:<10.2f}")

        # Select best model (lowest AIC)
        best_model_name, best_model, best_aic = min(models, key=lambda x: x[2])
        print(f"\nBest model: {best_model_name} (AIC = {best_aic:.2f})")
    else:
        print("No models successfully fitted")
        best_model = None
        best_model_name = None
else:
    print("Statsmodels not available - skipping model comparison")
    best_model = None
    best_model_name = None
```

```python
# Model Diagnostics - Residual Analysis
if statsmodels_available and best_model is not None:
    print("Model Diagnostics - Residual Analysis")
    print("=" * 60)

    # Choose best model for diagnostics, or fall back to Model 2
    diagnostic_model = best_model if best_model is not None else nb_model2
    diagnostic_name = best_model_name if best_model_name is not None else "Model 2"

    if diagnostic_model is not None:
        # Residuals analysis
        fitted_values = diagnostic_model.fittedvalues
        residuals = y - fitted_values

        # Standardized residuals for Q-Q plot
        standardized_residuals = residuals / np.sqrt(fitted_values)

        fig, axes = plt.subplots(2, 3, figsize=(18, 10))

        # Plot 1: Observed vs Fitted
        axes[0,0].scatter(fitted_values, y, alpha=0.7)
        axes[0,0].plot([y.min(), y.max()], [y.min(), y.max()], 'r--', alpha=0.8)
        axes[0,0].set_xlabel('Fitted Values')
        axes[0,0].set_ylabel('Observed Values')
        axes[0,0].set_title('Observed vs Fitted')
        axes[0,0].grid(True, alpha=0.3)

        # Plot 2: Residuals vs Fitted
        axes[0,1].scatter(fitted_values, residuals, alpha=0.7)
        axes[0,1].axhline(y=0, color='r', linestyle='--', alpha=0.8)
        axes[0,1].set_xlabel('Fitted Values')
        axes[0,1].set_ylabel('Residuals')
        axes[0,1].set_title('Residuals vs Fitted')
        axes[0,1].grid(True, alpha=0.3)

        # Plot 3: Q-Q plot of standardized residuals
        stats.probplot(standardized_residuals, dist="norm", plot=axes[0,2])
        axes[0,2].set_title('Q-Q Plot (Standardized Residuals)')
        axes[0,2].grid(True, alpha=0.3)

        # Plot 4: Time series of residuals
        axes[1,0].plot(modeling_data['year'], residuals, marker='o', markersize=4)
        axes[1,0].axhline(y=0, color='r', linestyle='--', alpha=0.8)
        axes[1,0].set_xlabel('Year')
        axes[1,0].set_ylabel('Residuals')
        axes[1,0].set_title('Residuals Over Time')
        axes[1,0].grid(True, alpha=0.3)

        # Plot 5: Fitted values over time
        axes[1,1].plot(modeling_data['year'], y, 'bo-', label='Observed', markersize=4)
        axes[1,1].plot(modeling_data['year'], fitted_values, 'ro-', label='Fitted', markersize=4)
        axes[1,1].set_xlabel('Year')
        axes[1,1].set_ylabel('DPCO Count')
        axes[1,1].set_title(f'{diagnostic_name} - Model Fit Over Time')
        axes[1,1].legend()
        axes[1,1].grid(True, alpha=0.3)

        # Plot 6: Histogram of standardized residuals
        axes[1,2].hist(standardized_residuals, bins=12, density=True, alpha=0.7, edgecolor='black')
        # Overlay normal distribution
        x_norm = np.linspace(standardized_residuals.min(), standardized_residuals.max(), 100)
        axes[1,2].plot(x_norm, stats.norm.pdf(x_norm, 0, 1), 'r-', alpha=0.8, label='Normal(0,1)')
        axes[1,2].set_xlabel('Standardized Residuals')
        axes[1,2].set_ylabel('Density')
        axes[1,2].set_title('Histogram of Standardized Residuals')
        axes[1,2].legend()
        axes[1,2].grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()
else:
    print("Cannot perform diagnostics - no valid model available")
```

```python
# Normality Tests and Goodness of Fit
if statsmodels_available and best_model is not None:
    print("Normality Tests and Goodness of Fit")
    print("=" * 60)

    diagnostic_model = best_model if best_model is not None else nb_model2
    diagnostic_name = best_model_name if best_model_name is not None else "Model 2"

    if diagnostic_model is not None:
        fitted_values = diagnostic_model.fittedvalues
        residuals = y - fitted_values
        standardized_residuals = residuals / np.sqrt(fitted_values)

        # Normality tests on standardized residuals
        shapiro_stat, shapiro_p = stats.shapiro(standardized_residuals)
        anderson_stat, anderson_crit, anderson_sig = stats.anderson(standardized_residuals, dist='norm')

        print(f"Normality tests on standardized residuals:")
        print(f"  Shapiro-Wilk test: W = {shapiro_stat:.4f}, p = {format_p_value(shapiro_p)}")
        if shapiro_p < 0.05:
            print(f"    → Reject normality assumption (p < 0.05)")
        else:
            print(f"    → Fail to reject normality assumption (p ≥ 0.05)")

        print(f"  Anderson-Darling test: A² = {anderson_stat:.4f}")
        for i, (crit, sig) in enumerate(zip(anderson_crit, anderson_sig)):
            if anderson_stat > crit:
                print(f"    → Reject normality at {sig}% significance level")
                break
        else:
            print(f"    → Fail to reject normality at all tested significance levels")

        # Goodness of fit statistics
        print(f"\nGoodness of fit ({diagnostic_name}):")
        print(f"  Pearson chi-squared: {diagnostic_model.pearson_chi2:.2f}")
        print(f"  Deviance: {diagnostic_model.deviance:.2f}")
        print(f"  Degrees of freedom: {diagnostic_model.df_resid}")

        # Check for overdispersion
        dispersion_ratio = diagnostic_model.pearson_chi2 / diagnostic_model.df_resid
        print(f"  Dispersion ratio: {dispersion_ratio:.3f}")
        if dispersion_ratio > 1.5:
            print("  → Overdispersion detected (ratio > 1.5)")
        elif dispersion_ratio < 0.7:
            print("  → Underdispersion detected (ratio < 0.7)")
        else:
            print("  → Dispersion appears adequate")
else:
    print("Cannot perform normality tests - no valid model available")
```
