```python
# Negative Binomial Regression for DPCO Counts - Setup and Data Preparation
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Define format_p_value function for consistent formatting
def format_p_value(p):
    """Format p-values consistently"""
    if p < 0.001:
        return "< 0.001"
    elif p < 0.01:
        return f"{p:.3f}"
    else:
        return f"{p:.3f}"

# Try importing statsmodels for negative binomial regression
try:
    import statsmodels.api as sm
    from scipy import stats
    statsmodels_available = True
except ImportError:
    print("Warning: statsmodels not available. Installing or using alternative approach.")
    statsmodels_available = False

print("Negative Binomial Regression Analysis for DPCO Counts")
print("=" * 60)

# Prepare data for modeling
modeling_data = yearly_counts.copy()

# Create time variables
modeling_data['years_since_start'] = modeling_data['year'] - modeling_data['year'].min()
modeling_data['year_centered'] = modeling_data['year'] - modeling_data['year'].mean()

# Create log-transformed mention predictors (add small constant to avoid log(0))
modeling_data['log_dis_count'] = np.log(modeling_data['dis_count'] + 1)
modeling_data['log_pnm_count'] = np.log(modeling_data['pnm_count'] + 1)
modeling_data['log_document_count'] = np.log(modeling_data['document_count'] + 1)

# Check for any issues with the data
print("Data preparation:")
print(f"Sample size: {len(modeling_data)}")
print(f"DPCO count range: {modeling_data['co_count'].min():.0f} to {modeling_data['co_count'].max():.0f}")
print(f"Mean DPCO count: {modeling_data['co_count'].mean():.2f}")
print(f"Variance DPCO count: {modeling_data['co_count'].var():.2f}")
print(f"Variance/Mean ratio: {modeling_data['co_count'].var()/modeling_data['co_count'].mean():.2f}")

# Prepare response variable
y = modeling_data['co_count'].astype(int)
```

```python
# Normality Tests and Goodness of Fit
if statsmodels_available and best_model is not None:
    print("Normality Tests and Goodness of Fit")
    print("=" * 60)

    diagnostic_model = best_model if best_model is not None else nb_model2
    diagnostic_name = best_model_name if best_model_name is not None else "Model 2"

    if diagnostic_model is not None:
        fitted_values = diagnostic_model.fittedvalues
        residuals = y - fitted_values
        standardized_residuals = residuals / np.sqrt(fitted_values)

        # Normality tests on standardized residuals
        shapiro_stat, shapiro_p = stats.shapiro(standardized_residuals)
        anderson_stat, anderson_crit, anderson_sig = stats.anderson(standardized_residuals, dist='norm')

        print(f"Normality tests on standardized residuals:")
        print(f"  Shapiro-Wilk test: W = {shapiro_stat:.4f}, p = {format_p_value(shapiro_p)}")
        if shapiro_p < 0.05:
            print(f"    → Reject normality assumption (p < 0.05)")
        else:
            print(f"    → Fail to reject normality assumption (p ≥ 0.05)")

        print(f"  Anderson-Darling test: A² = {anderson_stat:.4f}")
        for i, (crit, sig) in enumerate(zip(anderson_crit, anderson_sig)):
            if anderson_stat > crit:
                print(f"    → Reject normality at {sig}% significance level")
                break
        else:
            print(f"    → Fail to reject normality at all tested significance levels")

        # Goodness of fit statistics
        print(f"\nGoodness of fit ({diagnostic_name}):")
        print(f"  Pearson chi-squared: {diagnostic_model.pearson_chi2:.2f}")
        print(f"  Deviance: {diagnostic_model.deviance:.2f}")
        print(f"  Degrees of freedom: {diagnostic_model.df_resid}")

        # Check for overdispersion
        dispersion_ratio = diagnostic_model.pearson_chi2 / diagnostic_model.df_resid
        print(f"  Dispersion ratio: {dispersion_ratio:.3f}")
        if dispersion_ratio > 1.5:
            print("  → Overdispersion detected (ratio > 1.5)")
        elif dispersion_ratio < 0.7:
            print("  → Underdispersion detected (ratio < 0.7)")
        else:
            print("  → Dispersion appears adequate")
else:
    print("Cannot perform normality tests - no valid model available")
```

```python
# Preliminary Test for Autocorrelation
print("Preliminary Autocorrelation Analysis")
print("=" * 60)

# Test for autocorrelation in the DPCO count time series
if statsmodels_available:
    from statsmodels.stats.diagnostic import acorr_ljungbox
    from statsmodels.tsa.stattools import acf, pacf

    # Determine appropriate number of lags
    # Common guidelines: max lags = sqrt(N) or N/4, where N is sample size
    n_obs = len(y)
    max_lags_sqrt = int(np.sqrt(n_obs))
    max_lags_quarter = int(n_obs / 4)

    # Use conservative approach: min of sqrt(N) and 12 (common for annual data)
    max_lags = min(max_lags_sqrt, 12)

    print(f"Sample size: {n_obs} observations")
    print(f"√N = {max_lags_sqrt}, N/4 = {max_lags_quarter}")
    print(f"Using {max_lags} lags for testing (conservative approach)")
    print()
    print(f"Why {max_lags} lags was chosen:")
    print(f"  • With N=50 observations: √50 = {max_lags_sqrt:.1f} ≈ {max_lags} lags")
    print(f"  • Conservative upper bound for annual data: 12 lags")
    print(f"  • min({max_lags_sqrt}, 12) = {max_lags}")
    print(f"  • {max_lags} lags allows detection of cyclical patterns up to {max_lags} years")
    print(f"  • Avoids overfitting with too many parameters relative to sample size")
    print()

    # Ljung-Box test for autocorrelation
    try:
        ljung_box_result = acorr_ljungbox(y, lags=max_lags, return_df=True)

        print("Ljung-Box Test for Autocorrelation:")
        print(f"  Testing lags 1-{max_lags} for serial correlation in DPCO counts")
        print(f"  H0: No autocorrelation")
        print()

        # Check if any p-values are significant
        significant_lags = ljung_box_result[ljung_box_result['lb_pvalue'] < 0.05]

        if len(significant_lags) > 0:
            print(f"  ⚠️  Autocorrelation detected at lag(s): {list(significant_lags.index)}")
            print(f"  Minimum p-value: {ljung_box_result['lb_pvalue'].min():.4f}")
        else:
            print(f"  ✓ No significant autocorrelation detected")
            print(f"  Minimum p-value: {ljung_box_result['lb_pvalue'].min():.4f}")

        print(f"\nDetailed results:")
        print(ljung_box_result.round(4))

    except Exception as e:
        print(f"Ljung-Box test failed: {e}")

    # Plot autocorrelation functions
    try:
        # Use same number of lags for ACF/PACF plots
        plot_lags = min(15, max_lags + 3)  # Slightly more for visualization

        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

        # Autocorrelation function (ACF)
        acf_values = acf(y, nlags=plot_lags)
        ax1.stem(range(len(acf_values)), acf_values, basefmt=" ")
        ax1.axhline(y=0, color='black', linestyle='-', alpha=0.5)
        ax1.axhline(y=1.96/np.sqrt(len(y)), color='red', linestyle='--', alpha=0.7, label='95% CI')
        ax1.axhline(y=-1.96/np.sqrt(len(y)), color='red', linestyle='--', alpha=0.7)
        ax1.set_xlabel('Lag')
        ax1.set_ylabel('Autocorrelation')
        ax1.set_title(f'(a) Autocorrelation Function (ACF) - {plot_lags} lags')
        ax1.legend()
        ax1.grid(True, alpha=0.3)

        # Partial autocorrelation function (PACF)
        pacf_values = pacf(y, nlags=plot_lags)
        ax2.stem(range(len(pacf_values)), pacf_values, basefmt=" ")
        ax2.axhline(y=0, color='black', linestyle='-', alpha=0.5)
        ax2.axhline(y=1.96/np.sqrt(len(y)), color='red', linestyle='--', alpha=0.7, label='95% CI')
        ax2.axhline(y=-1.96/np.sqrt(len(y)), color='red', linestyle='--', alpha=0.7)
        ax2.set_xlabel('Lag')
        ax2.set_ylabel('Partial Autocorrelation')
        ax2.set_title(f'(b) Partial Autocorrelation Function (PACF) - {plot_lags} lags')
        ax2.legend()
        ax2.grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

        # Interpret autocorrelation results
        print(f"\nAutocorrelation Interpretation:")
        acf_significant = np.abs(acf_values[1:]) > 1.96/np.sqrt(len(y))
        pacf_significant = np.abs(pacf_values[1:]) > 1.96/np.sqrt(len(y))

        if np.any(acf_significant):
            sig_acf_lags = np.where(acf_significant)[0] + 1
            print(f"  ACF: Significant autocorrelation at lag(s) {list(sig_acf_lags)}")
        else:
            print(f"  ACF: No significant autocorrelation detected")

        if np.any(pacf_significant):
            sig_pacf_lags = np.where(pacf_significant)[0] + 1
            print(f"  PACF: Significant partial autocorrelation at lag(s) {list(sig_pacf_lags)}")
        else:
            print(f"  PACF: No significant partial autocorrelation detected")

    except Exception as e:
        print(f"ACF/PACF plotting failed: {e}")

    # Test for trend stationarity
    try:
        from statsmodels.tsa.stattools import adfuller

        print(f"\nAugmented Dickey-Fuller Test for Stationarity:")
        adf_result = adfuller(y, autolag='AIC')

        print(f"  ADF Statistic: {adf_result[0]:.4f}")
        print(f"  p-value: {format_p_value(adf_result[1])}")
        print(f"  Critical Values:")
        for key, value in adf_result[4].items():
            print(f"    {key}: {value:.4f}")

        if adf_result[1] < 0.05:
            print(f"  ✓ Series appears stationary (reject unit root hypothesis)")
        else:
            print(f"  ⚠️  Series may be non-stationary (fail to reject unit root hypothesis)")

    except Exception as e:
        print(f"ADF test failed: {e}")

    print(f"\nImplications for Regression Modeling:")
    if 'significant_lags' in locals() and len(significant_lags) > 0:
        print(f"  • Autocorrelation detected - consider:")
        print(f"    - Robust standard errors (HAC estimators)")
        print(f"    - AR/ARMA error terms")
        print(f"    - Generalized Least Squares (GLS)")
        print(f"    - Time series regression methods")
    else:
        print(f"  • No significant autocorrelation - standard regression should be adequate")
        print(f"  • However, monitor residuals for any remaining temporal patterns")

    print(f"\nNote on Lag Selection:")
    print(f"  • Common guidelines: max lags ≈ √N ({max_lags_sqrt}) or N/4 ({max_lags_quarter})")
    print(f"  • For annual data: often use 10-12 lags maximum")
    print(f"  • Too many lags can reduce test power with small samples")
    print(f"  • Used {max_lags} lags as conservative choice for N={n_obs}")

else:
    print("Statsmodels not available - skipping autocorrelation tests")
    print("Note: Autocorrelation testing is important for time series regression")
```

```python
# Quick Implementation: Robust Standard Errors
print("Quick Fix: Implementing Robust Standard Errors")
print("=" * 60)

if statsmodels_available:
    print("Re-running Model 2 with HAC (Newey-West) standard errors:")
    print()

    try:
        # Re-estimate Model 2 with robust covariance
        X2 = modeling_data[['years_since_start', 'log_dis_count', 'log_pnm_count']]
        X2 = sm.add_constant(X2)

        # Fit model
        nb_model2_robust = sm.GLM(y, X2, family=sm.families.NegativeBinomial()).fit()

        # Get robust covariance matrix (HAC with automatic lag selection)
        from statsmodels.stats.sandwich_covariance import cov_hac

        # Calculate HAC covariance matrix
        cov_hac_matrix = cov_hac(nb_model2_robust)

        # Get robust standard errors
        robust_se = np.sqrt(np.diag(cov_hac_matrix))

        # Calculate robust t-statistics and p-values
        robust_tstat = nb_model2_robust.params / robust_se
        robust_pvals = 2 * (1 - stats.norm.cdf(np.abs(robust_tstat)))

        print("COMPARISON: Standard vs. Robust Results")
        print("=" * 50)

        results_comparison = pd.DataFrame({
            'Coefficient': nb_model2_robust.params,
            'Std_SE': nb_model2_robust.bse,
            'Std_pvalue': nb_model2_robust.pvalues,
            'Robust_SE': robust_se,
            'Robust_pvalue': robust_pvals
        })

        print(results_comparison.round(6))
        print()

        # Focus on the time trend
        time_idx = X2.columns.get_loc('years_since_start')

        print("TIME TREND COMPARISON:")
        print(f"                    Standard    Robust")
        print(f"Standard Error:     {nb_model2_robust.bse.iloc[time_idx]:.6f}   {robust_se[time_idx]:.6f}")
        print(f"p-value:           {format_p_value(nb_model2_robust.pvalues.iloc[time_idx])}    {format_p_value(robust_pvals[time_idx])}")

        # Calculate robust confidence intervals
        alpha = 0.05
        z_crit = stats.norm.ppf(1 - alpha/2)

        time_coef = nb_model2_robust.params.iloc[time_idx]
        robust_ci_lower = time_coef - z_crit * robust_se[time_idx]
        robust_ci_upper = time_coef + z_crit * robust_se[time_idx]

        standard_ci_lower = time_coef - z_crit * nb_model2_robust.bse.iloc[time_idx]
        standard_ci_upper = time_coef + z_crit * nb_model2_robust.bse.iloc[time_idx]

        print(f"\n95% CONFIDENCE INTERVALS for time coefficient:")
        print(f"Standard CI:       [{standard_ci_lower:.6f}, {standard_ci_upper:.6f}]")
        print(f"Robust CI:         [{robust_ci_lower:.6f}, {robust_ci_upper:.6f}]")

        # Annual change rates with robust CI
        annual_rate = (np.exp(time_coef) - 1) * 100
        annual_rate_ci_lower = (np.exp(robust_ci_lower) - 1) * 100
        annual_rate_ci_upper = (np.exp(robust_ci_upper) - 1) * 100

        print(f"\nANNUAL CHANGE RATE:")
        print(f"Point estimate:    {annual_rate:+.2f}% per year")
        print(f"Robust 95% CI:     [{annual_rate_ci_lower:+.2f}%, {annual_rate_ci_upper:+.2f}%]")

        # Interpretation
        print(f"\nINTERPRETATION:")
        if robust_pvals[time_idx] < 0.05:
            print(f"✓ Time trend remains statistically significant with robust standard errors")
        else:
            print(f"⚠️  Time trend becomes non-significant with robust standard errors")

        print(f"• Robust standard errors are {robust_se[time_idx]/nb_model2_robust.bse.iloc[time_idx]:.2f}x larger than standard errors")
        print(f"• This reflects the additional uncertainty due to autocorrelation")
        print(f"• Use these robust results for your final conclusions")

    except Exception as e:
        print(f"Robust standard error calculation failed: {e}")
        print("Note: This is common with GLM models - consider using OLS as approximation")

else:
    print("Statsmodels not available - cannot compute robust standard errors")
```
