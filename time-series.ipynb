```python
# Populate each date
import pandas as pd
from datetime import datetime, timedelta

# Read data into dataframe (assuming CSV format with date and value columns)
# Replace 'your_file.csv' with actual filename
df = pd.read_csv('data/raw/time-series/time_series copy.csv')

# Convert date column to datetime (replace 'date' with actual column name)
df['date'] = pd.to_datetime(df['date'])

# Create complete date range from min to max date
date_range = pd.date_range(start=df['date'].min(), end=df['date'].max(), freq='D')

# Create dataframe with complete date range
complete_df = pd.DataFrame({'date': date_range})

# Merge with original data, filling missing dates with 0
df_filled = complete_df.merge(df, on='date', how='left').fillna(0)

df
```

```python
# Create pretty plots showing trends over time
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Set up the figure with four subplots sharing x-axis
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(15, 12), sharex=True)

# First subplot: CO Count
ax1.plot(df_filled['date'], df_filled['co_count'], label='CO Count', color='red', alpha=0.7)
ax1.set_ylabel('CO Count')
ax1.set_title('CO Count Over Time')
ax1.grid(True, alpha=0.3)

# Second subplot: Disease Count
ax2.plot(df_filled['date'], df_filled['dis_count'], label='Disease Count', color='blue', alpha=0.7)
ax2.set_ylabel('Disease Count')
ax2.set_title('Disease Count Over Time')
ax2.grid(True, alpha=0.3)

# Third subplot: PNM Count
ax3.plot(df_filled['date'], df_filled['pnm_count'], label='PNM Count', color='green', alpha=0.7)
ax3.set_ylabel('PNM Count')
ax3.set_title('PNM Count Over Time')
ax3.grid(True, alpha=0.3)

# Fourth subplot: Document Count Histogram
ax4.plot(df_filled['date'], df_filled['document_count'], color='orange', alpha=0.8, linewidth=0.5)
ax4.fill_between(df_filled['date'], df_filled['document_count'], alpha=0.3, color='orange')
ax4.set_ylabel('Abstract Count')
ax4.set_xlabel('Date')
ax4.set_title('Abstrac Count Distribution Over Time')
ax4.grid(True, alpha=0.3)

# Format x-axis dates
ax4.xaxis.set_major_locator(mdates.YearLocator(10))
ax4.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax4.xaxis.set_minor_locator(mdates.YearLocator(5))

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()
```

```python
# Calculate rates per abstract and create plots
import numpy as np

# Calculate rates per abstract (avoid division by zero)
df_rates = df_filled.copy()
df_rates['co_rate'] = np.where(df_rates['document_count'] > 0,
                               df_rates['co_count'] / df_rates['document_count'], 0)
df_rates['dis_rate'] = np.where(df_rates['document_count'] > 0,
                                df_rates['dis_count'] / df_rates['document_count'], 0)
df_rates['pnm_rate'] = np.where(df_rates['document_count'] > 0,
                                df_rates['pnm_count'] / df_rates['document_count'], 0)

# Set up the figure with four subplots sharing x-axis
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(15, 12), sharex=True)

# First subplot: CO Rate per Abstract
ax1.plot(df_rates['date'], df_rates['co_rate'], color='red', alpha=0.7)
ax1.set_ylabel('CO Rate per Abstract')
ax1.set_title('CO Rate per Abstract Over Time')
ax1.grid(True, alpha=0.3)

# Second subplot: Disease Rate per Abstract
ax2.plot(df_rates['date'], df_rates['dis_rate'], color='blue', alpha=0.7)
ax2.set_ylabel('Disease Rate per Abstract')
ax2.set_title('Disease Rate per Abstract Over Time')
ax2.grid(True, alpha=0.3)

# Third subplot: PNM Rate per Abstract
ax3.plot(df_rates['date'], df_rates['pnm_rate'], color='green', alpha=0.7)
ax3.set_ylabel('PNM Rate per Abstract')
ax3.set_title('PNM Rate per Abstract Over Time')
ax3.grid(True, alpha=0.3)

# Fourth subplot: Document Count for Reference
ax4.plot(df_rates['date'], df_rates['document_count'], color='orange', alpha=0.8, linewidth=0.5)
ax4.fill_between(df_rates['date'], df_rates['document_count'], alpha=0.3, color='orange')
ax4.set_ylabel('Abstract Count')
ax4.set_xlabel('Date')
ax4.set_title('Abstract Count Distribution Over Time (Reference)')
ax4.grid(True, alpha=0.3)

# Format x-axis dates
ax4.xaxis.set_major_locator(mdates.YearLocator(10))
ax4.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax4.xaxis.set_minor_locator(mdates.YearLocator(5))

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()
```

```python
# Filter data for 1975 to 2024-12-31 and create focused plots
df_filtered = df_rates[(df_rates['date'] >= '1975-01-01') & (df_rates['date'] <= '2024-12-31')].copy()

# Set up the figure with four subplots sharing x-axis
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(15, 12), sharex=True)

# First subplot: CO Rate per Abstract
ax1.plot(df_filtered['date'], df_filtered['co_rate'], color='red', alpha=0.7)
ax1.set_ylabel('CO Rate per Abstract')
ax1.set_title('CO Rate per Abstract Over Time (1975-2024)')
ax1.grid(True, alpha=0.3)

# Second subplot: Disease Rate per Abstract
ax2.plot(df_filtered['date'], df_filtered['dis_rate'], color='blue', alpha=0.7)
ax2.set_ylabel('Disease Rate per Abstract')
ax2.set_title('Disease Rate per Abstract Over Time (1975-2024)')
ax2.grid(True, alpha=0.3)

# Third subplot: PNM Rate per Abstract
ax3.plot(df_filtered['date'], df_filtered['pnm_rate'], color='green', alpha=0.7)
ax3.set_ylabel('PNM Rate per Abstract')
ax3.set_title('PNM Rate per Abstract Over Time (1975-2024)')
ax3.grid(True, alpha=0.3)

# Fourth subplot: Document Count for Reference
ax4.plot(df_filtered['date'], df_filtered['document_count'], color='orange', alpha=0.8, linewidth=0.5)
ax4.fill_between(df_filtered['date'], df_filtered['document_count'], alpha=0.3, color='orange')
ax4.set_ylabel('Abstract Count')
ax4.set_xlabel('Date')
ax4.set_title('Abstract Count Distribution Over Time (1975-2024)')
ax4.grid(True, alpha=0.3)

# Format x-axis dates
ax4.xaxis.set_major_locator(mdates.YearLocator(5))
ax4.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax4.xaxis.set_minor_locator(mdates.YearLocator(1))

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()
```

```python
# Aggregate data by year and create yearly plots
df_yearly = df_filtered.copy()
df_yearly['year'] = df_yearly['date'].dt.year

# Group by year and sum the counts
yearly_agg = df_yearly.groupby('year').agg({
    'document_count': 'sum',
    'co_count': 'sum',
    'dis_count': 'sum',
    'pnm_count': 'sum'
}).reset_index()

# Calculate yearly rates
yearly_agg['co_rate'] = np.where(yearly_agg['document_count'] > 0,
                                yearly_agg['co_count'] / yearly_agg['document_count'], 0)
yearly_agg['dis_rate'] = np.where(yearly_agg['document_count'] > 0,
                                 yearly_agg['dis_count'] / yearly_agg['document_count'], 0)
yearly_agg['pnm_rate'] = np.where(yearly_agg['document_count'] > 0,
                                 yearly_agg['pnm_count'] / yearly_agg['document_count'], 0)

# Set up the figure with four subplots sharing x-axis
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(15, 12), sharex=True)

# First subplot: CO Rate per Abstract (Yearly)
ax1.plot(yearly_agg['year'], yearly_agg['co_rate'], color='red', alpha=0.7, marker='o', markersize=3)
ax1.set_ylabel('CO Rate per Abstract')
ax1.set_title('Yearly CO Rate per Abstract (1975-2024)')
ax1.grid(True, alpha=0.3)

# Second subplot: Disease Rate per Abstract (Yearly)
ax2.plot(yearly_agg['year'], yearly_agg['dis_rate'], color='blue', alpha=0.7, marker='o', markersize=3)
ax2.set_ylabel('Disease Rate per Abstract')
ax2.set_title('Yearly Disease Rate per Abstract (1975-2024)')
ax2.grid(True, alpha=0.3)

# Third subplot: PNM Rate per Abstract (Yearly)
ax3.plot(yearly_agg['year'], yearly_agg['pnm_rate'], color='green', alpha=0.7, marker='o', markersize=3)
ax3.set_ylabel('PNM Rate per Abstract')
ax3.set_title('Yearly PNM Rate per Abstract (1975-2024)')
ax3.grid(True, alpha=0.3)

# Fourth subplot: Document Count (Yearly)
ax4.bar(yearly_agg['year'], yearly_agg['document_count'], color='orange', alpha=0.7, width=0.8)
ax4.set_ylabel('Total Abstract Count')
ax4.set_xlabel('Year')
ax4.set_title('Yearly Abstract Count (1975-2024)')
ax4.grid(True, alpha=0.3)

# Format x-axis
ax4.set_xlim(1975, 2024)

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()

# Display summary statistics
print("Yearly aggregated data summary:")
print(yearly_agg.describe())
```

```python
# Aggregate data by month and create monthly plots
df_monthly = df_filtered.copy()
df_monthly['year_month'] = df_monthly['date'].dt.to_period('M')

# Group by month and sum the counts
monthly_agg = df_monthly.groupby('year_month').agg({
    'document_count': 'sum',
    'co_count': 'sum',
    'dis_count': 'sum',
    'pnm_count': 'sum'
}).reset_index()

# Convert period to datetime for plotting
monthly_agg['date'] = monthly_agg['year_month'].dt.to_timestamp()

# Calculate monthly rates
monthly_agg['co_rate'] = np.where(monthly_agg['document_count'] > 0,
                                 monthly_agg['co_count'] / monthly_agg['document_count'], 0)
monthly_agg['dis_rate'] = np.where(monthly_agg['document_count'] > 0,
                                  monthly_agg['dis_count'] / monthly_agg['document_count'], 0)
monthly_agg['pnm_rate'] = np.where(monthly_agg['document_count'] > 0,
                                  monthly_agg['pnm_count'] / monthly_agg['document_count'], 0)

# Set up the figure with four subplots sharing x-axis
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(15, 12), sharex=True)

# First subplot: CO Rate per Abstract (Monthly)
ax1.plot(monthly_agg['date'], monthly_agg['co_rate'], color='red', alpha=0.7, linewidth=0.8)
ax1.set_ylabel('CO Rate per Abstract')
ax1.set_title('Monthly CO Rate per Abstract (1975-2024)')
ax1.grid(True, alpha=0.3)

# Second subplot: Disease Rate per Abstract (Monthly)
ax2.plot(monthly_agg['date'], monthly_agg['dis_rate'], color='blue', alpha=0.7, linewidth=0.8)
ax2.set_ylabel('Disease Rate per Abstract')
ax2.set_title('Monthly Disease Rate per Abstract (1975-2024)')
ax2.grid(True, alpha=0.3)

# Third subplot: PNM Rate per Abstract (Monthly)
ax3.plot(monthly_agg['date'], monthly_agg['pnm_rate'], color='green', alpha=0.7, linewidth=0.8)
ax3.set_ylabel('PNM Rate per Abstract')
ax3.set_title('Monthly PNM Rate per Abstract (1975-2024)')
ax3.grid(True, alpha=0.3)

# Fourth subplot: Document Count (Monthly)
ax4.plot(monthly_agg['date'], monthly_agg['document_count'], color='orange', alpha=0.8, linewidth=0.8)
ax4.fill_between(monthly_agg['date'], monthly_agg['document_count'], alpha=0.3, color='orange')
ax4.set_ylabel('Total Abstract Count')
ax4.set_xlabel('Date')
ax4.set_title('Monthly Abstract Count (1975-2024)')
ax4.grid(True, alpha=0.3)

# Format x-axis dates
ax4.xaxis.set_major_locator(mdates.YearLocator(5))
ax4.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax4.xaxis.set_minor_locator(mdates.YearLocator(1))

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()

# Display summary statistics
print("Monthly aggregated data summary:")
print(monthly_agg[['document_count', 'co_rate', 'dis_rate', 'pnm_rate']].describe())
```

```python
# Create histogram plots at different time aggregations (no date filtering)
# Use the original df_filled data for all time periods

# Prepare aggregations for the full dataset
df_full_monthly = df_filled.copy()
df_full_monthly['year_month'] = df_full_monthly['date'].dt.to_period('M')
monthly_full_agg = df_full_monthly.groupby('year_month').agg({'document_count': 'sum'}).reset_index()
monthly_full_agg['date'] = monthly_full_agg['year_month'].dt.to_timestamp()

df_full_yearly = df_filled.copy()
df_full_yearly['year'] = df_full_yearly['date'].dt.year
yearly_full_agg = df_full_yearly.groupby('year').agg({'document_count': 'sum'}).reset_index()

# Set up the figure with three subplots sharing x-axis
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(15, 17), sharex=True)


# First subplot: Daily document count (full time series)
ax1.plot(df_filled['date'], df_filled['document_count'], color='orange', alpha=0.8, linewidth=0.3)
ax1.fill_between(df_filled['date'], df_filled['document_count'], alpha=0.2, color='orange')
ax1.set_ylabel('Daily Abstract Count')
ax1.set_ylim(bottom=0)
ax1.grid(True, alpha=0.3)

# Second subplot: Monthly document count
ax2.plot(monthly_full_agg['date'], monthly_full_agg['document_count'], color='blue', alpha=0.8, linewidth=0.8)
ax2.fill_between(monthly_full_agg['date'], monthly_full_agg['document_count'], alpha=0.3, color='blue')
ax2.set_ylabel('Monthly Abstract Count')
ax2.set_ylim(bottom=0)
ax2.grid(True, alpha=0.3)

# Third subplot: Yearly document count - convert years to datetime for consistent x-axis
yearly_dates = pd.to_datetime(yearly_full_agg['year'], format='%Y')
ax3.bar(yearly_dates, yearly_full_agg['document_count'], color='green', alpha=0.7, width=300)
ax3.set_ylabel('Yearly Abstract Count')
ax3.set_xlabel('Date')
ax3.set_ylim(bottom=0)
ax3.grid(True, alpha=0.3)

# Format shared x-axis dates with 5-year intervals and limit to 2024-12-31
ax3.xaxis.set_major_locator(mdates.YearLocator(5))
ax3.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax3.xaxis.set_minor_locator(mdates.YearLocator(1))
ax3.set_xlim(left=pd.to_datetime('1794-01-01'), right=pd.to_datetime('2024-12-31'))

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()

print("Summary statistics:")
print(f"Daily data points: {len(df_filled)}")
print(f"Monthly data points: {len(monthly_full_agg)}")
print(f"Yearly data points: {len(yearly_full_agg)}")
```

```python
# Create yearly aggregated plot with histograms and rate overlay lines (1975-2024)
# Filter data for the specified time range
df_filtered_yearly = df_rates[(df_rates['date'] >= '1975-01-01') & (df_rates['date'] <= '2024-12-31')].copy()
df_filtered_yearly['year'] = df_filtered_yearly['date'].dt.year

# Group by year and sum the counts
yearly_counts = df_filtered_yearly.groupby('year').agg({
    'document_count': 'sum',
    'dis_count': 'sum',
    'pnm_count': 'sum',
    'co_count': 'sum'
}).reset_index()

# Calculate rates per abstract
yearly_counts['dis_rate'] = yearly_counts['dis_count'] / yearly_counts['document_count']
yearly_counts['pnm_rate'] = yearly_counts['pnm_count'] / yearly_counts['document_count']
yearly_counts['co_rate'] = yearly_counts['co_count'] / yearly_counts['document_count']

# Custom formatter for thousand separators with thin space
from matplotlib.ticker import FuncFormatter
def thousands_formatter(x, pos):
    return f"{x:,.0f}".replace(',', ' ')

# Set up the figure with four subplots - each with independent x-axis
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(15, 14.4))

# First subplot: DIS Count histogram with rate overlay
ax1_twin = ax1.twinx()
ax1.bar(yearly_counts['year'], yearly_counts['dis_count'], color='red', alpha=0.7, width=0.8)
ax1_twin.plot(yearly_counts['year'], yearly_counts['dis_rate'], color='darkgrey', linewidth=2, marker='o', markersize=3)
ax1.set_ylabel('DIS Count')
ax1_twin.set_ylabel('DIS Rate per Abstract')
ax1.set_ylim(bottom=0)
ax1_twin.set_ylim(0, 5)
ax1.grid(True, alpha=0.3, which='both')
ax1.yaxis.set_major_locator(plt.MultipleLocator(500000))
ax1.yaxis.set_minor_locator(plt.MultipleLocator(250000))
ax1.xaxis.set_major_locator(plt.MultipleLocator(5))
ax1.xaxis.set_minor_locator(plt.MultipleLocator(1))
ax1.set_xlim(1975, 2024)
# Format with thousand separators
ax1.yaxis.set_major_formatter(FuncFormatter(thousands_formatter))
ax1_twin.yaxis.set_major_formatter(plt.ScalarFormatter(useOffset=False))

# Second subplot: PNM Count histogram with rate overlay
ax2_twin = ax2.twinx()
ax2.bar(yearly_counts['year'], yearly_counts['pnm_count'], color='orange', alpha=0.7, width=0.8)
ax2_twin.plot(yearly_counts['year'], yearly_counts['pnm_rate'], color='darkgrey', linewidth=2, marker='o', markersize=3)
ax2.set_ylabel('PNM Count')
ax2_twin.set_ylabel('PNM Rate per Abstract')
ax2.set_ylim(0, 250000)
ax2_twin.set_ylim(0, 0.25)
ax2.grid(True, alpha=0.3, which='both')
ax2.yaxis.set_major_locator(plt.MultipleLocator(50000))
ax2.yaxis.set_minor_locator(plt.MultipleLocator(25000))
ax2.xaxis.set_major_locator(plt.MultipleLocator(5))
ax2.xaxis.set_minor_locator(plt.MultipleLocator(1))
ax2.set_xlim(1975, 2024)
# Format with thousand separators
ax2.yaxis.set_major_formatter(FuncFormatter(thousands_formatter))
ax2_twin.yaxis.set_major_formatter(plt.ScalarFormatter(useOffset=False))

# Third subplot: DPCO Count histogram with rate overlay
ax3_twin = ax3.twinx()
ax3.bar(yearly_counts['year'], yearly_counts['co_count'], color='purple', alpha=0.7, width=0.8)
ax3_twin.plot(yearly_counts['year'], yearly_counts['co_rate'], color='darkgrey', linewidth=2, marker='o', markersize=3)
ax3.set_ylabel('DPCO Count')
ax3_twin.set_ylabel('DPCO Rate per Abstract')
ax3.set_ylim(0, 50000)
ax3_twin.set_ylim(0, 0.1)
ax3.grid(True, alpha=0.3, which='both')
ax3.yaxis.set_major_locator(plt.MultipleLocator(10000))
ax3.yaxis.set_minor_locator(plt.MultipleLocator(5000))
ax3.xaxis.set_major_locator(plt.MultipleLocator(5))
ax3.xaxis.set_minor_locator(plt.MultipleLocator(1))
ax3.set_xlim(1975, 2024)
# Format with thousand separators
ax3.yaxis.set_major_formatter(FuncFormatter(thousands_formatter))
ax3_twin.yaxis.set_major_formatter(plt.ScalarFormatter(useOffset=False))

# Fourth subplot: Document Count histogram
ax4.bar(yearly_counts['year'], yearly_counts['document_count'], color='lightsteelblue', alpha=0.7, width=0.8)
ax4.set_ylabel('Abstract Count')
ax4.set_xlabel('Year')
ax4.set_ylim(bottom=0)
ax4.grid(True, alpha=0.3, which='both')
ax4.yaxis.set_major_locator(plt.MultipleLocator(100000))
ax4.yaxis.set_minor_locator(plt.MultipleLocator(50000))
ax4.xaxis.set_major_locator(plt.MultipleLocator(5))
ax4.xaxis.set_minor_locator(plt.MultipleLocator(1))
ax4.set_xlim(1975, 2024)
# Format with thousand separators
ax4.yaxis.set_major_formatter(FuncFormatter(thousands_formatter))

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()

# Display summary statistics
print("Yearly counts summary (1975-2024):")
print(yearly_counts.describe())
```
